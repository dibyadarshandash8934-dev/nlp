```python
# ======================================================================
# Step 1: Setup — install and import libraries
# ======================================================================
# Install split-folders: splits a folder of class-subfolders into train/val/test
!pip install split-folders -q

import os
import zipfile
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import splitfolders
from sklearn.metrics import confusion_matrix, classification_report

# Colab's built-in helper for uploading/downloading files in the browser
from google.colab import files

print("TensorFlow version:", tf.__version__)
print("GPU available:", tf.config.list_physical_devices('GPU'))

# ======================================================================
# Step 2: Upload your dataset
# ======================================================================
# Upload the dataset zip file from your computer
uploaded = files.upload()

# Assumes a single zip file was uploaded
zip_filename = list(uploaded.keys())[0]

# Extract it into a folder called "raw_dataset"
extract_path = "raw_dataset"
with zipfile.ZipFile(zip_filename, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print(f"Extracted '{zip_filename}' into '{extract_path}/'")

# If the zip contained one single top-level folder, step into it automatically
contents = [c for c in os.listdir(extract_path) if not c.startswith('.')]
if len(contents) == 1 and os.path.isdir(os.path.join(extract_path, contents[0])):
    extract_path = os.path.join(extract_path, contents[0])

print("Classes found:", os.listdir(extract_path))

# ======================================================================
# Step 3: Organize into train / validation / test sets
# ======================================================================
output_path = "dataset_split"

splitfolders.ratio(
    extract_path,             # source folder, one subfolder per class
    output=output_path,       # where the split dataset will be created
    seed=42,                  # for reproducible splits
    ratio=(0.7, 0.15, 0.15)   # train, val, test
)

train_dir = os.path.join(output_path, "train")
val_dir = os.path.join(output_path, "val")
test_dir = os.path.join(output_path, "test")

class_names = sorted(os.listdir(train_dir))
num_classes = len(class_names)

print("Classes:", class_names)
print("Number of classes:", num_classes)

# ======================================================================
# Step 4: Preprocess and augment the images
# ======================================================================
IMG_SIZE = (224, 224)
BATCH_SIZE = 32

# Load images straight from folders — Keras automatically infers class labels
# from the subfolder names.
train_ds = keras.utils.image_dataset_from_directory(
    train_dir, image_size=IMG_SIZE, batch_size=BATCH_SIZE, label_mode='categorical'
)
val_ds = keras.utils.image_dataset_from_directory(
    val_dir, image_size=IMG_SIZE, batch_size=BATCH_SIZE, label_mode='categorical'
)
test_ds = keras.utils.image_dataset_from_directory(
    test_dir, image_size=IMG_SIZE, batch_size=BATCH_SIZE, label_mode='categorical', shuffle=False
)

# --- Normalization: rescales pixel values from 0-255 down to 0-1 ---
normalization_layer = layers.Rescaling(1./255)

# --- Data augmentation: randomly perturbs images, applied to TRAINING data only ---
data_augmentation = keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),
    layers.RandomContrast(0.1),
], name="data_augmentation")

def prepare(ds, augment=False):
    ds = ds.map(lambda x, y: (normalization_layer(x), y), num_parallel_calls=tf.data.AUTOTUNE)
    if augment:
        ds = ds.map(lambda x, y: (data_augmentation(x, training=True), y),
                    num_parallel_calls=tf.data.AUTOTUNE)
    return ds.prefetch(tf.data.AUTOTUNE)   # prefetch overlaps data loading with training for speed

train_ds_prepped = prepare(train_ds, augment=True)
val_ds_prepped = prepare(val_ds, augment=False)
test_ds_prepped = prepare(test_ds, augment=False)

# ======================================================================
# Step 5: Build a simple CNN model
# ======================================================================
def build_cnn(input_shape=(224, 224, 3), num_classes=num_classes):
    model = keras.Sequential([
        layers.Input(shape=input_shape),

        # Block 1: learns simple features like edges and colors
        layers.Conv2D(32, 3, activation='relu', padding='same'),
        layers.MaxPooling2D(),

        # Block 2: learns slightly more complex patterns
        layers.Conv2D(64, 3, activation='relu', padding='same'),
        layers.MaxPooling2D(),

        # Block 3
        layers.Conv2D(128, 3, activation='relu', padding='same'),
        layers.MaxPooling2D(),

        # Block 4
        layers.Conv2D(128, 3, activation='relu', padding='same'),
        layers.MaxPooling2D(),

        # Classifier head
        layers.Flatten(),
        layers.Dropout(0.5),                          # reduces overfitting
        layers.Dense(128, activation='relu'),
        layers.Dense(num_classes, activation='softmax')  # one probability per class
    ], name="simple_cnn")
    return model

model = build_cnn()

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()

# ======================================================================
# Step 6: Train the model
# ======================================================================
EPOCHS = 20

callbacks = [
    keras.callbacks.EarlyStopping(monitor='val_loss', patience=4, restore_best_weights=True),
    keras.callbacks.ModelCheckpoint('best_model.keras', monitor='val_accuracy', save_best_only=True)
]

history = model.fit(
    train_ds_prepped,
    validation_data=val_ds_prepped,
    epochs=EPOCHS,
    callbacks=callbacks
)

# Final metrics from the last completed epoch
print(f"Final Training Accuracy:   {history.history['accuracy'][-1]:.4f}")
print(f"Final Validation Accuracy: {history.history['val_accuracy'][-1]:.4f}")
print(f"Final Training Loss:       {history.history['loss'][-1]:.4f}")
print(f"Final Validation Loss:     {history.history['val_loss'][-1]:.4f}")

# ======================================================================
# Step 7: Plot accuracy and loss curves
# ======================================================================
acc = history.history['accuracy']
val_acc = history.history['val_accuracy']
loss = history.history['loss']
val_loss = history.history['val_loss']
epochs_range = range(1, len(acc) + 1)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(epochs_range, acc, label='Training Accuracy')
plt.plot(epochs_range, val_acc, label='Validation Accuracy')
plt.legend(loc='lower right')
plt.title('Accuracy over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')

plt.subplot(1, 2, 2)
plt.plot(epochs_range, loss, label='Training Loss')
plt.plot(epochs_range, val_loss, label='Validation Loss')
plt.legend(loc='upper right')
plt.title('Loss over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')

plt.tight_layout()
plt.show()

# ======================================================================
# Step 8: Evaluate on the held-out test set
# ======================================================================
# Overall accuracy/loss on the test set
test_loss, test_acc = model.evaluate(test_ds_prepped)
print(f"Test Accuracy: {test_acc:.4f}")
print(f"Test Loss:     {test_loss:.4f}")

# Collect true labels and predicted labels across the whole test set
y_true, y_pred = [], []
for images, labels in test_ds_prepped:
    preds = model.predict(images, verbose=0)
    y_true.extend(np.argmax(labels.numpy(), axis=1))
    y_pred.extend(np.argmax(preds, axis=1))

# Confusion matrix: rows = actual class, columns = predicted class
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(7, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()

# Classification report: precision, recall, F1-score for every class
print(classification_report(y_true, y_pred, target_names=class_names))

# ======================================================================
# Step 9 & 10: Predict on a brand-new image
# ======================================================================
# Upload new image(s) to classify
new_uploads = files.upload()

for fname in new_uploads.keys():
    # Load and resize to match the model's expected input size
    img = keras.utils.load_img(fname, target_size=IMG_SIZE)
    img_array = keras.utils.img_to_array(img) / 255.0    # normalize, same as training
    img_array = np.expand_dims(img_array, axis=0)         # add batch dimension -> shape (1, 224, 224, 3)

    # Predict
    predictions = model.predict(img_array, verbose=0)[0]
    predicted_index = int(np.argmax(predictions))
    predicted_class = class_names[predicted_index]
    confidence = float(predictions[predicted_index]) * 100

    # Display the image with its prediction
    plt.figure(figsize=(4, 4))
    plt.imshow(img)
    plt.axis('off')
    plt.title(f"{predicted_class} ({confidence:.2f}%)")
    plt.show()

    print(f"File: {fname}")
    print(f"Predicted class: {predicted_class}")
    print(f"Confidence: {confidence:.2f}%")
    print("-" * 40)

# ======================================================================
# Step 11: Save the trained model
# ======================================================================
model_filename = "image_classifier_model.keras"
model.save(model_filename)
print(f"Model saved as {model_filename}")

# Download it to your computer
files.download(model_filename)
```
